```python
import uproot
```

```python
import awkward as ak
```

```python
import numpy as np
import pandas as pd
```

```python
sig_file = uproot.open("data_reco_raw/sig_noMuConst.root")
```

```python
bkg_file = uproot.open("data_reco_raw/bkg_noMuConst_new.root")
```

```python
sig_tree = sig_file["analysis"]
```

```python
bkg_tree = bkg_file["analysis"]
```

```python
# 1. Define the exact fj_track columns we want to keep
allowed_fj_tracks = set()
for i in range(1, 31):
    allowed_fj_tracks.update([
        f"fj_track{i}Pt", 
        f"fj_track{i}Eta", 
        f"fj_track{i}Phi", 
        #f"fj_track{i}TruthOrigin",
        #f"fj_track{i}TruthType",
    ])

# 2. Filter the tree's keys before loading
all_keys = bkg_tree.keys()
keys_to_keep = []
for key in all_keys:
    if not key.startswith("fj_track"):
        keys_to_keep.append(key)       # Keep all non-fj_track columns
    elif key in allowed_fj_tracks:
        keys_to_keep.append(key)       # Only keep the fj_track columns that match our list

# 3. Load the data using ONLY the filtered keys
events = bkg_tree.arrays(keys_to_keep)

# 4. Create the boolean mask
mask1 = ak.any(events["muon_fatjet_dr"] < 1.2, axis=1)

mask2 = ak.any(events["fatJetM"] > 20, axis=1)

combined_mask = mask1 & mask2
filtered_events = events[combined_mask]

# 6. Safely convert the filtered data to a Pandas DataFrame
df_bkg = ak.to_dataframe(filtered_events)

df_bkg.to_parquet("df_bkg_4V_noMu.parquet", index=False)
```

```python
# 1. Define the exact fj_track columns we want to keep
allowed_fj_tracks = set()
for i in range(1, 31):
    allowed_fj_tracks.update([
        f"fj_track{i}Pt", 
        f"fj_track{i}Eta", 
        f"fj_track{i}Phi", 
        f"fj_track{i}TruthOrigin",
        f"fj_track{i}TruthType",
    ])

# 2. Filter the tree's keys before loading
all_keys = sig_tree.keys()
keys_to_keep = []
for key in all_keys:
    if not key.startswith("fj_track"):
        keys_to_keep.append(key)       # Keep all non-fj_track columns
    elif key in allowed_fj_tracks:
        keys_to_keep.append(key)       # Only keep the fj_track columns that match our list

# 3. Load the data using ONLY the filtered keys
events = sig_tree.arrays(keys_to_keep)

# 4. Create the boolean mask
mask = ak.any(events["muon_fatjet_dr"] < 1.5, axis=1)

# 5. Apply the mask to filter the events
filtered_events = events[mask]

# 6. Safely convert the filtered data to a Pandas DataFrame
df_sig = ak.to_dataframe(filtered_events)

print(df_sig.head())
```
